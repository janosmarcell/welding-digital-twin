```python
import cv2
import numpy as np
from scipy import ndimage
from matplotlib import pyplot as plt

# Kép betöltése
img = cv2.imread('c:/users/hj/downloads/eszes2/1/panorama.jpg', cv2.IMREAD_GRAYSCALE)

# Zajcsökkentés Gauss-szűrővel
img = cv2.GaussianBlur(img, (3, 3), 0)

# Roberts operátor maszkok
roberts_cross_v = np.array([[1, 0], [0, -1]])
roberts_cross_h = np.array([[0, 1], [-1, 0]])

# Konvolúció a Roberts maszkokkal
edges_x = ndimage.convolve(img, roberts_cross_v)
edges_y = ndimage.convolve(img, roberts_cross_h)

# Élek kombinálása
edges = np.sqrt(np.square(edges_x) + np.square(edges_y))

# Eredmény megjelenítése
plt.subplot(1, 2, 1), plt.imshow(img, cmap='gray')
plt.title('Eredeti kép'), plt.xticks([]), plt.yticks([])
plt.subplot(1, 2, 2), plt.imshow(edges, cmap='gray')
plt.title('Roberts élek'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
import numpy as np
import cv2
from matplotlib import pyplot as plt

# Betöltünk egy képet
image = cv2.imread('c:/users/hj/downloads/eszes2/1/panorama.jpg', cv2.IMREAD_GRAYSCALE)

# Definiáljuk a Roberts operátor kerneljeit
Gx = np.array([[1, 0], [0, -1]])
Gy = np.array([[0, 1], [-1, 0]])

# Alkalmazzuk a kerneljeit a szürkeárnyalatos képre
edge_x = cv2.filter2D(image, -1, Gx)
edge_y = cv2.filter2D(image, -1, Gy)

# Számoljuk ki a gradiens nagyságát
edge_magnitude = np.sqrt(np.square(edge_x) + np.square(edge_y))

# Normalizáljuk a nagyságot a [0, 255] tartományra
edge_magnitude = (edge_magnitude / np.max(edge_magnitude)) * 255

# Konvertáljuk a nagyságot unsigned 8-bit típusra
edge_magnitude = edge_magnitude.astype(np.uint8)

# Megjelenítjük az eredeti és az éldetektált képet
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Eredeti kép')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(edge_magnitude, cmap='gray')
plt.title('Éldetektált kép')
plt.axis('off')

plt.show()
```
