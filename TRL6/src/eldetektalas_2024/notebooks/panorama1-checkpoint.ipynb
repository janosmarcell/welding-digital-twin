```python
import cv2
import numpy as np

# Kép betöltése
image = cv2.imread('c:/users/hj/downloads/eszes2/1/panorama.jpg')
#gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Eredeti kép méretének lekérdezése
height, width = image.shape[:2]

# Kép átméretezése felére
resized_img = cv2.resize(image, (width // 3, height // 3))

# Átméretezett kép megjelenítése
cv2.imshow('Resized Image', resized_img)
cv2.waitKey(0)
cv2.destroyAllWindows()


```

```python
# Gaussian Blur alkalmazása a zaj csökkentésére
blurred = cv2.GaussianBlur(resized_img, (5, 5), 0)

# Canny élkereső alkalmazása
edges = cv2.Canny(blurred, 100, 150)

# Kontúrok keresése
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Kontúrok rajzolása az eredeti képre
cv2.drawContours(image, contours, -1, (0, 255, 0), 2)

# Eredmény megjelenítése
cv2.imshow('Weld Inspection', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Kontúrok középpontjainak meghatározása
centers = []
for contour in contours:
    M = cv2.moments(contour)
    if M['m00'] != 0:
        cx = int(M['m10'] / M['m00'])
        cy = int(M['m01'] / M['m00'])
        centers.append((cx, cy))

# Távolságok kiszámítása
distances = []
for i in range(len(centers)):
    for j in range(i + 1, len(centers)):
        dist = np.sqrt((centers[i][0] - centers[j][0])**2 + (centers[i][1] - centers[j][1])**2)
        distances.append(dist)
```

```python
# Eredmény megjelenítése
cv2.imshow('Weld Inspection', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
