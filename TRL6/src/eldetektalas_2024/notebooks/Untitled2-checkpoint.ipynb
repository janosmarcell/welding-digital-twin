```python
import cv2

# Kép beolvasása és előfeldolgozása:  Olvassuk be a képet és alakítsuk át szürkeárnyalatossá. Alkalmazzunk elmosást a zaj csökkentése érdekében
# Kép beolvasása
image = cv2.imread('c:/users/hj/downloads/eszes2/1/image_3.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Elmosás
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
```

```python
# Éldetektálás, canny éldetektáló a varratok éleinek megtalálásához.
edges = cv2.Canny(blurred, 50, 150)
```

```python
# Küszöbölés  Küszöbölés és morfológiai transzformációk:
# Alkalmazzunk küszöbölést a varratok háttértől való elkülönítéséhez.
# Használjunk morfológiai transzformációkat a varratok kiemeléséhez.
_, thresholded = cv2.threshold(edges, 127, 255, cv2.THRESH_BINARY_INV)

# Morfológiai transzformációk
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
opened = cv2.morphologyEx(thresholded, cv2.MORPH_OPEN, kernel)
```

```python
# Kontúrok megtalálása
# Keressük meg a kontúrokat a feldolgozott képen.
# Szűrjük ki a releváns kontúrokat terület és arány alapján.

contours, _ = cv2.findContours(opened, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Kontúrok szűrése
min_area = 500
max_aspect_ratio = 3
filtered_contours = []

for contour in contours:
    area = cv2.contourArea(contour)
    x, y, w, h = cv2.boundingRect(contour)
    aspect_ratio = w / h
    if area > min_area and aspect_ratio < max_aspect_ratio:
        filtered_contours.append(contour)
```

```python
# Kontúrok rajzolása és szélesség meghatározása: Rajzoljuk meg a szűrt kontúrokat az eredeti képen.
# Határozzuk meg a varrat szélességét a kontúrok alapján.
# Kontúrok rajzolása
# Kontúrok rajzolása és méretek kiíratása
for contour in filtered_contours:
    x, y, w, h = cv2.boundingRect(contour)
    cv2.drawContours(image, [contour], 0, (0, 255, 0), 2)
    cv2.putText(image, f'{w}x{h} px', (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

# Eredmény megjelenítése
cv2.imshow('Objects with Sizes', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
